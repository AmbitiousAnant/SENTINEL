## 1. Keystroke Kinematic Features
1. **Flight-Time / Inter-Key Delay ($IKD$):** Time elapsed between consecutive key-presses. Elevated in psychomotor retardation (depression).
2. **Backspace Revision Ratio ($BR$):** Ratio of backspace/delete events to total keystrokes. Elevated during cognitive indecision, anxiety, and rumination.


```python
# Baseline Healthy Session (Fluid typing, low corrections)
normal_session = KeystrokeSession(
    flight_times=[120.0, 140.0, 110.0, 135.0, 125.0, 130.0, 115.0, 140.0],
    backspace_count=1,
    total_keystrokes=40
)

# Depressive / Psychomotor Retarded Session (Sluggish typing, hesitation)
depressive_session = KeystrokeSession(
    flight_times=[350.0, 420.0, 380.0, 510.0, 460.0, 390.0, 480.0],
    backspace_count=12,
    total_keystrokes=30
)

m_normal = engine.analyze_session(normal_session)
m_depressive = engine.analyze_session(depressive_session)

print("--- NORMAL SESSION ---")
print(f"Mean Flight-Time:    {m_normal.mean_flight_time_ms:.1f} ms (Z: {m_normal.flight_time_z:+.2f} sigma)")
print(f"Revision Ratio:      {m_normal.backspace_ratio * 100:.1f}% (Z: {m_normal.backspace_ratio_z:+.2f} sigma)")
print(f"Psychomotor Anomaly: {m_normal.is_psychomotor_anomaly}")

print("\n--- DEPRESSIVE / RUMINATIVE SESSION ---")
print(f"Mean Flight-Time:    {m_depressive.mean_flight_time_ms:.1f} ms (Z: {m_depressive.flight_time_z:+.2f} sigma)")
print(f"Revision Ratio:      {m_depressive.backspace_ratio * 100:.1f}% (Z: {m_depressive.backspace_ratio_z:+.2f} sigma)")
print(f"Psychomotor Anomaly: {m_depressive.is_psychomotor_anomaly}")
```
